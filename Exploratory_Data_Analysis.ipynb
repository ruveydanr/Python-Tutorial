```python
columns_to_group =sorted(['2022 Population',
       '2020 Population', '2015 Population', '2010 Population',
       '2000 Population', '1990 Population', '1980 Population',
       '1970 Population'], key = lambda x: int(x.split()[0]))
df2 = df.groupby('Continent')[columns_to_group].mean(numeric_only= True).sort_values(by='2022 Population', ascending =False)
df2
```

```python
df3= df2.transpose()
df3
```
